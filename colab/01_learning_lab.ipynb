```python
import numpy as np
import pandas as pd
from lab.learning_engine import LearningEngine

# Replace this cell with your OHLCV loader. Required columns: open, high, low, close, volume.
# Example: df = pd.read_csv('/content/BTCUSDT.csv', parse_dates=['timestamp']).set_index('timestamp')

def trend_signal(data, p):
    fast = data.close.rolling(p['fast']).mean()
    slow = data.close.rolling(p['slow']).mean()
    return (fast > slow).astype(float)

def rsi_signal(data, p):
    delta = data.close.diff()
    up = delta.clip(lower=0).ewm(alpha=1/p['rsi'], adjust=False).mean()
    down = (-delta.clip(upper=0)).ewm(alpha=1/p['rsi'], adjust=False).mean()
    rsi = 100 - 100/(1 + up/down.replace(0, np.nan))
    return (rsi < p['entry']).astype(float)
```
