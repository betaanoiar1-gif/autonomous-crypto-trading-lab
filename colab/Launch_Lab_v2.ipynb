# Autonomous Crypto Trading Lab v2

Fresh launcher. No paid API key required.


```python
import pathlib, shutil, subprocess, sys, os
REPO_URL = 'https://github.com/betaanoiar1-gif/autonomous-crypto-trading-lab.git'
REPO_DIR = '/content/autonomous_crypto_trading_lab_v2'
print('=== Autonomous Crypto Trading Lab v2 ===')
if pathlib.Path(REPO_DIR).exists():
    print('Removing previous v2 workspace...')
    shutil.rmtree(REPO_DIR)
print('Cloning fresh project...')
clone = subprocess.run(['git','clone','--depth','1',REPO_URL,REPO_DIR], text=True, capture_output=True)
if clone.returncode != 0:
    print(clone.stdout)
    print(clone.stderr)
    raise RuntimeError(f'Git clone failed with code {clone.returncode}.')
print('Installing dependencies...')
install = subprocess.run([sys.executable,'-m','pip','install','-q','-r',f'{REPO_DIR}/requirements.txt'], text=True, capture_output=True)
if install.returncode != 0:
    print(install.stdout)
    print(install.stderr)
    raise RuntimeError(f'Dependency installation failed with code {install.returncode}.')
sys.path.insert(0, REPO_DIR)
os.environ['AGENT_PROVIDER'] = 'local'
print('Loading local research agent...')
from lab.local_agent import LocalAgent
agent = LocalAgent()
health = agent.healthcheck()
print(f"Local agent: {'OK' if health['ok'] else 'FAILED'} | model={health['model']} | device={health.get('device','unknown')}")
if not health['ok']:
    raise RuntimeError('Local research agent failed to load.')
print('Launching autonomous research...')
from lab.orchestrator import run
run(agent=agent)
print('=== Launch complete ===')
```
